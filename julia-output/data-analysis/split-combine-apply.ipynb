```julia
iris = RDatasets.dataset("datasets","iris")
```

```julia
simpsons_chars = CSV.read("../../assets/datasets/simpsons_characters.csv", DataFrame)
```

```julia
simpsons_episodes = CSV.read("../../assets/datasets/simpsons_episodes.csv", DataFrame)
```

```julia
@chain iris begin
  groupby(:Species)
end
```

### Same using DataFramesMeta

```julia
@chain iris begin
  @groupby(:Species)
  @combine begin
    :num = length(:SepalLength)
    :mean_SepalLength = mean(:SepalLength)
    :std_SepalLength = std(:SepalLength)
    :max_PetalWidth = maximum(:PetalWidth)
  end
end
```

```julia
@chain iris begin
  @by :Species begin
    :num = length(:SepalLength)
    :mean_SepalLength = mean(:SepalLength)
    :std_SepalLength = std(:SepalLength)
    :max_PetalWidth = maximum(:PetalWidth)
  end
end
```

```julia
@chain simpsons_episodes begin
  @by :original_air_year begin
    :num_per_year = length(:number_in_season)
  end
end
```
