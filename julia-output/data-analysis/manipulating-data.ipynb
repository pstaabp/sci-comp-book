```julia
using DataFrames, Chain, Statistics
```

```julia
data = DataFrame(
				  A = 1:2:13,
				  B = ["M", "F", "F", "M", "X", "F", "M"],
				  C = [3.0, 2.5, pi, -2.3, 1/3, 56, 100],
				  D = [(-1)^n//n for n=1:7]
				)
```

```julia
data[!,[:A,:D]]
```

```julia
select(data,:A,:D)
```

```julia
@chain data begin
  select(Not(:B))
end
```

```julia
@chain data begin
  select(Between(:B,:D))
end
```

```julia
@chain data begin
  subset(:A => a -> a .< 10)
end
```

```julia
@chain data begin
  subset(:B => b -> b .== "F")
end
```

```julia
@chain data begin
  subset([:C, :D] => (c, d) -> c .* d .> 2)
end
```

```julia
@chain data begin
  subset([:C, :D] => ByRow((c, d) -> c * d > 2))
end
```

```julia
@chain data begin
  subset([:C, :D] => (c, d) -> @. c * d > 2)
end
```

```julia
@chain data begin
  select(:A => a -> a.^2)
end
```

```julia
@chain data begin
  transform(:A => (a -> a.^2) => :Asq)
end
```

```julia
@chain data begin
  select([:C, :D] => ((c,d) -> @. c*d) => :prod)
end
```

## Stacking and Unstacking

```julia
df = DataFrame(
    ID=[1, 2, 3, 4],
    Name=["Ralph", "Bart", "Milhouse", "Nelson"],
    Q1=[85, 90, 78, 65],
    Q2=[88, 92, 80, 70],
    Q3=[90, 85, 82, 55]
)
```

```julia
tall_df = @chain df begin
  stack([:Q1, :Q2 ,:Q3], [:ID, :Name])
end
```

```julia
springfield_6k = DataFrame(
    Bib = [101, 102, 103, 205],
    Group = ["Moe's Bar", "Patty/Selma/Marge" , "Power Plant", "Elementary School"],
    Runner1 = [21.5, 24.0, 19.8, 23.8],      # times in minutes
    Runner2 = [18.5, 24.5, 30.5, 15.5],
    Runner3 = [21.0, 28.5, 32.0, 22.5]
)
```

```julia
unstack(tall_df)
```

```julia
springfield_sales = DataFrame(
  name = ["Moe's", "Kwik-E-Mart", "Krusty Burger", "Moe's", "Kwik-E-Mart", "Krusty Burger", "Moe's", "Kwik-E-Mart", "Krusty Burger","Moe's", "Kwik-E-Mart", "Krusty Burger"],
  month = ["Jan","Jan","Jan", "Feb", "Feb", "Feb", "Mar","Mar", "Mar", "Apr", "Apr", "Apr"],
  sales = [28,47,96,19,43,88,25,50,79,30,50,75] 
)
```

```julia
unstack(springfield_sales, :month, :sales)
```

```julia
using RDatasets
```

```julia
iris = RDatasets.dataset("datasets","iris")
```

```julia
@chain iris begin
  groupby(:Species)
  combine(:SepalLength => mean)
end
```

```julia
@chain iris begin
  groupby(:Species)
  combine(nrow)
end
```

```julia
@chain iris begin
  groupby(:Species)
  combine(nrow, :SepalLength => mean, :SepalWidth => std, :PetalWidth => maximum)
end
```
