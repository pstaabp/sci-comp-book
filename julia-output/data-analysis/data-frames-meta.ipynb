```julia
using DataFramesMeta
```

```julia
data = DataFrame(
		A = 1:2:13,
		B = ["M", "F", "F", "M", "X", "F", "M"],
		C = [3.0, 2.5, pi, -2.3, 1/3, 56, 100],
		D = [(-1)^n//n for n=1:7]
)
```

```julia
@chain data begin
  @select(:A, :D)
end
```

```julia
@chain data begin
  subset(:A => a-> a .< 10)
end
```

```julia
@chain data begin
  @subset(:A .< 10)
end
```

```julia
data.A .< 10
```

```julia
@chain data begin
  @subset(:C .* :D .> 2)
end
```

```julia
@chain data begin
  @subset(@. :C * :D > 2)
end
```

```julia
@chain data begin
  @transform(:Asq = :A.^2)
end
```

```julia
@chain data begin
  @select(:Asq = :A.^2)
end
```

```julia
@chain data begin
  @select(:A, :B, :Asq = :A.^2)
end
```

```julia
@chain data begin
  @select(:prod = :C .* :D)
end
```

```julia
@orderby(data,:C)
```

```julia
@chain data begin
  @orderby(:C)
end
```

```julia
@chain data begin
  @orderby(:B)
end
```

```julia
@chain data begin
  @orderby(-:C)
end
```

```julia
@chain data begin
  @orderby(:A .% 3)
end
```

```julia
data2 = DataFrame(
  B = ["M", "M", "F", "X"],
  Q = 1:4
)
```

```julia
innerjoin(data, data2, on = :B)
```

```julia
using Statistics
@combine(data, :mean_C = mean(:C), :mean_D = mean(:D), :std_C = std(:C), :n = length(:A))
```
