```julia
using DataFramesMeta
```

```julia
data = DataFrame(
		A = 1:2:13,
		B = ["M", "F", "F", "M", "X", "F", "M"],
		C = [3.0, 2.5, pi, -2.3, 1/3, 56, 100],
		D = [(-1)^n//n for n=1:7]
)
```

```julia
@chain data begin
  @select(:A, :D)
end
```

```julia
@chain data begin
  subset(:A => a-> a .< 10)
end
```
