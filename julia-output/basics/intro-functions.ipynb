```julia
function fact(n::Integer)
  local prod=1
  for i=1:n
    prod *= i
  end
  prod
end
```

```julia
fact(5), fact(6)
```

```julia
function factr(n::Integer)
  if n == 0
    return 1
  else
    return n*factr(n-1)
  end
end
```

```julia
factr(5), factr(6), factr(0)
```

```julia
function fibonacci(n::Integer)
  if n == 1 || n == 2
    return 1
  else
    return fibonacci(n-1) + fibonacci(n-2)
  end
end
```

```julia
for i=1:8
  println("fibonacci($i) = ", fibonacci(i))
end
```
