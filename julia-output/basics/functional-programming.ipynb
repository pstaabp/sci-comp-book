```julia
numPos(arr) = reduce((n,val) -> val > 0 ? n+1 : n, arr, init=0)
```

```julia
numPos([-3,4,-1,11,0,4])
```

```julia
numHi(v) = reduce((n,val) -> val == "hi" ? n+1 : n, v, init = 0)
```

```julia
numHi(["hi","bye","hi","hello"] )
```

```julia
numPos(values::Vector{Int}) = reduce(values, init = 0) do n, val
    begin
        pos = n
        if val > 0
            pos += 1
        end
        pos
    end
end
```

```julia
numPos([-3,5,8,-2,11])
```

```julia
findMax(v) = reduce(v, init = -Inf) do val, max 
  begin
    if val > max 
      return val 
    else
      return max
    end
  end
end
```

```julia
findMax(v) = reduce(v, init = -Inf) do val, max 
  begin
    val > max ? val : max 
  end
end
```

```julia
findMax([14,3, 17, -2])
```
