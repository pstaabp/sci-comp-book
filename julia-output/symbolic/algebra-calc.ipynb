```julia
@variables a[0:2] b[0:2] x
```

```julia
p1 = a[0] + a[1]*x + a[2]*x^2
p2 = b[0] + b[1]*x + b[2]*x^2
```

```julia
prod = expand(p1*p2)
```

```julia
anti = integrate(prod, x)
```

```julia
expand(15*(substitute(anti, x=> 1) - substitute(anti , x => -1)))
```
